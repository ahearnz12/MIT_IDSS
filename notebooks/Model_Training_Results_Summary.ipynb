## 🔬 Methodology and Theoretical Foundation

### **Data Preprocessing and Setup**
- **Train-Test Split**: Data split using Surprise library's `train_test_split` function
- **Rating Scale**: Implicit 5-point rating scale (1-5)
- **Evaluation Threshold**: 3.5 rating threshold for binary relevance (liked/not liked)
- **Cross-Validation**: 3-fold cross-validation for hyperparameter tuning

### **Evaluation Metrics Used**
- **Precision@k**: Proportion of recommended items that are relevant
  - Formula: `TP / (TP + FP)` where k=10
- **Recall@k**: Proportion of relevant items that are recommended  
  - Formula: `TP / (TP + FN)` where k=10
- **F1-Score@k**: Harmonic mean of precision and recall
  - Formula: `2 * (Precision * Recall) / (Precision + Recall)`
- **RMSE**: Root Mean Square Error for rating prediction accuracy
  - Formula: `√(Σ(predicted - actual)² / n)`

## 🤖 Models Implemented and Results

### **Collaborative Filtering Approaches**

#### **Memory-Based Collaborative Filtering (KNN)**
**Theory**: Finds similar users or items based on historical interactions and uses their preferences to make recommendations.

**Mathematical Foundation**:
- **Cosine Similarity**: `sim(u,v) = cos(θ) = (u·v) / (||u|| ||v||)`
- **Mean Squared Difference**: `sim(u,v) = 1 / (1 + MSD(u,v))`
- **Prediction Formula**: `r̂ᵤᵢ = μᵤ + Σ(sim(u,v) × (rᵥᵢ - μᵥ)) / Σ|sim(u,v)|`

#### **Model-Based Collaborative Filtering (SVD)**
**Theory**: Decomposes the user-item interaction matrix into lower-dimensional latent factor matrices.

**Mathematical Foundation**:
- **Matrix Factorization**: `R ≈ P × Qᵀ`
- **Objective Function**: `min Σ(rᵤᵢ - qᵢᵀpᵤ)² + λ(||qᵢ||² + ||pᵤ||²)`
- **Gradient Descent Updates**:
  - `pᵤf ← pᵤf + γ(eᵤᵢ × qᵢf - λ × pᵤf)`
  - `qᵢf ← qᵢf + γ(eᵤᵢ × pᵤf - λ × qᵢf)`

### 1. User-User Similarity-based Collaborative Filtering (KNNBasic)

- **Prediction Method**: Weighted average of similar users' ratings

### 3. Matrix Factorization (SVD)

**Algorithm Overview**:
- **Type**: Model-based collaborative filtering
- **Approach**: Learn latent factors for users and items through matrix decomposition
- **Implementation**: Surprise library's `SVD` algorithm
- **Optimization**: Stochastic Gradient Descent (SGD) with regularization
- **Key Features**: Handles sparse data well, learns hidden patterns, scalable

**Mathematical Details**:
- **Latent Factors**: Users and items represented in k-dimensional space
- **Regularization**: L2 regularization to prevent overfitting
- **Bias Terms**: Global, user, and item biases incorporated
- **Prediction Formula**: `r̂ᵤᵢ = μ + bᵤ + bᵢ + qᵢᵀpᵤ`

#### **Baseline SVD Model**
- **Configuration**: Default SVD parameters, random_state=1
- **Performance Metrics**:
  - 🎯 **Precision@10**: ~0.85
  - 🔍 **Recall@10**: ~0.88
  - ⚖️ **F1-Score@10**: ~0.87

#### **Optimized SVD Model**
- **Hyperparameter Tuning**: GridSearchCV with 3-fold cross-validation
- **Parameter Grid**:
  - `n_epochs`: [5, 10, 20]
  - `lr_all`: [0.002, 0.005, 0.01]
  - `reg_all`: [0.02, 0.05, 0.1]
- **Best Parameters**: n_epochs=20, lr_all=0.01, reg_all=0.1
- **Best RMSE**: ~0.90
- **Performance Metrics**:
  - 🎯 **Precision@10**: ~0.88
  - 🔍 **Recall@10**: ~0.84
  - ⚖️ **F1-Score@10**: ~0.86

### 🔧 **Algorithmic Complexity and Computational Considerations**:

| Algorithm | Training Complexity | Prediction Complexity | Memory Usage | Scalability |
|-----------|-------------------|---------------------|--------------|-------------|
| **User-KNN** | O(m²n) | O(km) | O(m²) | Poor for large user base |
| **Item-KNN** | O(mn²) | O(kn) | O(n²) | Poor for large item catalog |
| **SVD** | O(k³ + kmn) | O(k) | O(k(m+n)) | Excellent |

**Where**: m = users, n = items, k = latent factors/neighbors

### 🎯 **Model-Specific Strengths and Limitations**:

**KNN-Based Models**:
- ✅ **Strengths**: Intuitive, explainable, no training phase
- ❌ **Limitations**: Scalability issues, sparse data problems, cold start
- 🎯 **Best For**: Small-medium datasets, explainable recommendations

**SVD Matrix Factorization**:
- ✅ **Strengths**: Handles sparsity well, scalable, learns latent patterns
- ❌ **Limitations**: Black box, requires parameter tuning, needs sufficient data
- 🎯 **Best For**: Large datasets, production systems, cold start scenarios
